```python
import pandas as pd
```

```python
df = pd.read_csv("/Users/sergeybudygin/Desktop/micha/itog.csv") 
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(12, 6))

plt.plot(df['Явка 2012'], label='Явка 2012')
plt.plot(df['Явка 2018'], label='Явка 2018')
plt.plot(df['Поддержка 2012'], label='Поддержка 2012')
plt.plot(df['Поддержка 2018'], label='Поддержка 2018')

plt.legend()
plt.title('Явка и поддержка Путина на выборах')
plt.xlabel('УИК (индекс)')
plt.ylabel('Доля')
plt.grid(True)
plt.show()
```

```python
import plotly.express as px

# Преобразуем координаты
df[['lat', 'lon']] = df['coords_x'].str.strip("()").str.split(",", expand=True).astype(float)

fig = px.scatter_mapbox(
    df,
    lat="lat",
    lon="lon",
    color="Поддержка 2018",
    size="Явка 2018",
    hover_name="uik",
    mapbox_style="carto-positron",
    zoom=10,
    title="Поддержка и явка на выборах 2018"
)
fig
```
